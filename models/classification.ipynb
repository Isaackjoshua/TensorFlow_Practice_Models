DATA PREPROCESSING

Unlike models like decision trees or random forests Neural nets are scale sensitive, thus its hard for them to handle values on random scale, therefore we normalize the data

```python
X_train, X_test = X_train/255.0, X_test/255.0
```

```python
X_train[0]
```

```python
y_train
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input, Dropout

model = Sequential()
model.add(Input((28, 28)))
model.add(Flatten())
model.add(Dense(128, activation='relu')) 
model.add(Dropout(0.2))
model.add(Dense(10))
 
```

```python
nonsense_prediction = model(X_train[:1]).numpy()
nonsense_prediction
```

```python
tf.nn.softmax(nonsense_prediction).numpy().sum()
```

```python
from tensorflow.keras.losses import SparseCategoricalCrossentropy
loss_fn = SparseCategoricalCrossentropy(from_logits = True)
loss_fn(y_train[:1], nonsense_prediction)
```

NOW WE ARE GOING TO COMPILE THE MODEL AND TRAIN IT;
this means that we are going to define an optimizer, specify the loss function, we choose metrics to take a look at, and make the model ready for training.

```python
from tensorflow.keras.optimizers import Adam
optimizer = Adam(learning_rate = 0.01)
model.compile(optimizer=optimizer, loss=loss_fn, metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10)
```

```python
model.evaluate(X_test, y_test)
```

NOTE THAT : Hyperparameter tuning is done on the third portion of data called the validation dataset for optimal performance
